```python
import pandas as pd
from dash import Dash, dcc, html
import plotly.express as px

# Load Dataset
df = pd.read_excel("ApexPlanet_DataAnalytics_Dataset.xlsx")

# Convert date column
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# ---------------- KPIs ----------------
total_sales = df["Total_Sales"].sum()
avg_price = df["Unit_Price"].mean()
total_qty = df["Quantity"].sum()
avg_age = df["Age"].mean()

# ---------------- Charts ----------------

# Sales Trend
line_fig = px.line(
    df.groupby("Order_Date", as_index=False)["Total_Sales"].sum(),
    x="Order_Date",
    y="Total_Sales",
    title="Total Sales Trend"
)

# Sales by Gender
pie_gender = px.pie(
    df,
    names="Gender",
    values="Total_Sales",
    hole=0.5,
    title="Sales by Gender"
)

# Sales by Category
pie_category = px.pie(
    df,
    names="Category",
    values="Total_Sales",
    title="Sales by Category"
)

# Treemap
tree = px.treemap(
    df,
    path=["Product"],
    values="Total_Sales",
    title="Sales by Product"
)

# Bar Chart
bar = px.bar(
    df.groupby("Category", as_index=False)["Total_Sales"].sum(),
    x="Total_Sales",
    y="Category",
    orientation="h",
    title="Sales by Category"
)

# ---------------- Dashboard ----------------

app = Dash(__name__)

app.layout = html.Div([

    html.H1("Sales Dashboard", style={"textAlign":"center"}),

    html.Div([
        html.Div([
            html.H3(f"{total_sales:,.2f}"),
            html.P("Total Sales")
        ], style={"width":"23%","display":"inline-block","border":"1px solid gray","padding":"10px"}),

        html.Div([
            html.H3(f"{avg_price:,.2f}"),
            html.P("Average Unit Price")
        ], style={"width":"23%","display":"inline-block","border":"1px solid gray","padding":"10px"}),

        html.Div([
            html.H3(f"{total_qty:,.0f}"),
            html.P("Total Quantity")
        ], style={"width":"23%","display":"inline-block","border":"1px solid gray","padding":"10px"}),

        html.Div([
            html.H3(f"{avg_age:.2f}"),
            html.P("Average Age")
        ], style={"width":"23%","display":"inline-block","border":"1px solid gray","padding":"10px"}),
    ]),

    html.Br(),

    html.Div([
        dcc.Graph(figure=line_fig, style={"width":"48%","display":"inline-block"}),
        dcc.Graph(figure=pie_gender, style={"width":"48%","display":"inline-block"}),
    ]),

    html.Div([
        dcc.Graph(figure=pie_category, style={"width":"48%","display":"inline-block"}),
        dcc.Graph(figure=tree, style={"width":"48%","display":"inline-block"}),
    ]),

    html.Div([
        dcc.Graph(figure=bar)
    ])

])

if __name__ == "__main__":
    app.run(debug=True)
```
